```python
import numpy as np
```

```python
import pandas as pd

class BankingSystem:
    def __init__(self):
        self.accounts = pd.DataFrame(columns=['Account Number', 'Balance', 'Transactions'])

    def create_account(self, account_number, initial_balance):
        if (self.accounts['Account Number'] == account_number).any():
            print("Account already exists")
        else:
            history = [f"Initial deposit: {initial_balance}"]
            self.accounts.loc[len(self.accounts)] = [account_number, initial_balance, history]

    def deposit(self, account_number, amount):
        mask = self.accounts['Account Number'] == account_number
        if mask.any():
            current_balance = self.accounts.loc[mask, 'Balance'].values[0]
            self.accounts.loc[mask, 'Balance'] = current_balance + amount
            self.accounts.loc[mask, 'Transactions'].values[0].append(f"Deposit: +{amount}")
        else:
            print("Account not found")

    def withdraw(self, account_number, amount):
        mask = self.accounts['Account Number'] == account_number
        if mask.any():
            current_balance = self.accounts.loc[mask, 'Balance'].values[0]
            if current_balance >= amount:
                self.accounts.loc[mask, 'Balance'] = current_balance - amount
                self.accounts.loc[mask, 'Transactions'].values[0].append(f"Withdrawal: -{amount}")
            else:
                print("Insufficient funds")
        else:
            print("Account not found")

    def get_balance(self, account_number):
        mask = self.accounts['Account Number'] == account_number
        if mask.any():
            return self.accounts.loc[mask, 'Balance'].values[0]
        else:
            return None

    def get_transaction_history(self, account_number):
        mask = self.accounts['Account Number'] == account_number
        if mask.any():
            return self.accounts.loc[mask, 'Transactions'].values[0]
        else:
            return None
```

```python
# 1. Initialize the banking system
bank = BankingSystem()
```

```python
# 2. Create two accounts to verify indexing works properly
bank.create_account("ACC001", 1000)
bank.create_account("ACC002", 500)
```

```python
# 3. Perform deposits and withdrawals
bank.deposit("ACC001", 250)
bank.withdraw("ACC001", 100)
bank.withdraw("ACC002", 600)  # Should print "Insufficient funds"
```

```python
# 4.Check outcomes
print("ACC001 Balance:", bank.get_balance("ACC001"))
print("\nAll Accounts Table:")
print(bank.accounts)
```

```python
print("ACC001 Transaction History:")
print(bank.get_transaction_history("ACC001"))
```

```python
print("\nACC002 Transaction History:")
print(bank.get_transaction_history("ACC002"))
```

```python
print("\nACC002 Balance:", bank.get_balance("ACC002"))
```

```python
print("\nTesting invalid account calls:")
bank.deposit("ACC999", 500)                 # Should print "Account not found"
bank.withdraw("ACC999", 100)                # Should print "Account not found"
print("ACC999 Balance:", bank.get_balance("ACC999")) # Should print "None"
```

```python
# 5. Multiple sequential operations on a single account
bank.deposit("ACC002", 150)
bank.withdraw("ACC002", 200)
bank.deposit("ACC002", 50)
print("\nACC002 Final Balance (should be 500):", bank.get_balance("ACC002"))
print("ACC002 Updated Transaction History:")
print(bank.get_transaction_history("ACC002"))
```
